```python
df3 = pd.read_csv("customers.csv")
```

```python
df3.head()
```

```python
columns_df3 = df3.columns.tolist()
# Remove customer_name from its current position
columns_df3.remove('customer_name')
# Find the index of 'customer_id'
customer_id_index = columns_df3.index('customer_id')
# Insert 'customer_name' after 'customer_id'
columns_df3.insert(customer_id_index + 1, 'customer_name')

df3 = df3[columns_df3]
display(df3.head())
```

```python
df3['customer_name'] = df3['first_name'] + " " + df3['last_name']
```

```python
df3.drop(['first_name', 'last_name'], axis=1, inplace=True)
```

```python
df3['phone'] = df3['phone'].fillna("Not Avalable")
```

```python
df3.isnull().sum()
```
